# Getting Results and Peaks

Query processed results with `GetResults` and their peak data with `GetPeaks`, using `GetFieldNames` to find the available fields.

```python
import os

from dotenv import load_dotenv

from OptiHPLCHandler import EmpowerHandler

load_dotenv("../.env")
EMPOWER_API_ADDRESS = os.getenv("EMPOWER_API_ADDRESS_PRD")
EXAMPLE_PROJECT = os.getenv("EXAMPLE_PROJECT")
```

```python
if EMPOWER_API_ADDRESS:
    handler = EmpowerHandler(
        project=EXAMPLE_PROJECT,
        address=EMPOWER_API_ADDRESS,
        allow_login_without_context_manager=True,
    )
    handler.login()
```

## Available fields

```python
result_field_names = [field["name"] for field in handler.GetFieldNames("Result")]
print(result_field_names)
```

## Result by ID

`GetResult` returns every available field of one result.

```python
result_id = 1870  # Replace with a result ID from your project
result = handler.GetResult(result_id)
```

## Results by search

`search_field`, `operator` and `condition` work like the ViewFilter in the Empower Project window, as in `GetSampleSets`. The search field is added to `returned_fields` if missing.

- Operators: `handler.allowed_result_operators` (`EQ`, `NEQ`, `LT`, `LTE`, `GT`, `GTE`)
- Ordering: `handler.allowed_order_by_options` (`OrderNone`, `Ascending`, `Descending`)
- Dates can be given as e.g. `15 April 2025` or `15/04/2025`; use `"*"` to ignore the condition

Results are paged, by default `page_number=1` and `page_size=50`.

```python
searched_results = handler.GetResults(
    search_field="DateAcquired",
    operator="GT",
    condition="01/01/2025",
    returned_fields=["ChannelName", "SampleName"],
    order_by="Descending",
    page_size=10,
)
```

## Peaks

```python
peak_field_names = [field["name"] for field in handler.GetFieldNames("Peak")]
peak_field_names
```

```python
# Default peak fields, one dict per peak
peaks = handler.GetPeaks(
    result_ids=[result_id]
)  # returned_fields = None returns all fields.
peaks
```

```python
# Selected peak fields as returned by Empower; "id" is the result each peak belongs to
raw_peaks = handler.GetPeaksRaw(
    result_ids=[result_id], returned_fields=["RetentionTime", "Area"]
)
raw_peaks
```
