# Getting Sample Sets

Query sample set data in a project with `GetSampleSets`, using `GetFieldNames` to find the available fields.

```python
import os

from dotenv import load_dotenv

from OptiHPLCHandler import EmpowerHandler

load_dotenv("../.env")
EMPOWER_API_ADDRESS = os.getenv("EMPOWER_API_ADDRESS_PRD")
EXAMPLE_PROJECT = os.getenv("EXAMPLE_PROJECT")
```

```python
if EMPOWER_API_ADDRESS:
    handler = EmpowerHandler(
        project=EXAMPLE_PROJECT,
        address=EMPOWER_API_ADDRESS,
        allow_login_without_context_manager=True,
    )
    handler.login()
```

## Available fields

Allowed field types are listed in `handler.allowed_field_types`.

```python
print(handler.allowed_field_types)

sample_set_fields = handler.GetFieldNames("SampleSet")
for field in sample_set_fields:
    print(field)
```

```python
sample_set_fields = handler.GetFieldNames("SampleSetLine")
for field in sample_set_fields:
    print(field)
```

## Get all sample sets

```python
# Getting list of sample set methods names
list_sample_set_method_name = (
    handler.GetSampleSetMethodList()
)  # Maybe change to sampleset method names?
sample_set_method_name = (
    list_sample_set_method_name[0] if list_sample_set_method_name else None
)
```

## Filter and sort

`search_field`, `operator` and `condition` work like the ViewFilter in the Empower Project window.

- Operators: `handler.allowed_result_operators` (`EQ`, `NEQ`, `LT`, `LTE`, `GT`, `GTE`)
- Ordering: `handler.allowed_order_by_options` (`OrderNone`, `Ascending`, `Descending`)
- Dates can be given as `15/04/2025`

```python
print(handler.allowed_result_operators)
print(handler.allowed_order_by_options)
```

```python
# search_field is added to returned_fields automatically if missing
returned_fields = ["SampleSetId", "SampleSetName"]
recent_sample_sets = handler.GetSampleSets(
    search_field="SystemCreateDate",
    operator="GT",
    condition="01/10/2026",
    returned_fields=returned_fields,
    order_by="Descending",
)
recent_sample_sets
```

```python
named_sample_sets = handler.GetSampleSets(
    search_field="SampleSetName",
    operator="EQ",
    condition="clean",
    returned_fields=returned_fields,
)
named_sample_sets
```

```python
# All fields for a named sample set; a list, as sample set names are not unique
handler.GetSampleSet("clean")
```
